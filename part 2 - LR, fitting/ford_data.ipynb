```python
X = df.drop('price', axis=1)
y = df['price']
```

```python
X_one_hot_encoded = pd.get_dummies(X, columns=['model', 'transmission', 'fuelType'], drop_first=True)
```

```python
X_one_hot_encoded = X_one_hot_encoded.astype(int)
X_one_hot_encoded.head(n=10)
```

```python
from scipy.stats import chi2_contingency

alpha = 0.05

cat_features = ['year', 'mileage', 'engineSize', 'transmission', 'fuelType', 'tax']

X['charges_bin'] = pd.qcut(y, q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contigency = pd.crosstab(X[col], X['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contigency)
    decision = 'Reject Null (Keep the feature)' if p_val < alpha else 'Fail to Reject Null (Drop the feature)'
    chi2_results[col] = {
        'chi2_stat': chi2_stat,
        'p_value': p_val,
        'decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python
X.drop(['charges_bin'], axis=1, inplace=True)
```

```python
X.head()
```

```python
from sklearn.preprocessing import LabelEncoder
encode = LabelEncoder()

columns = ['model', 'transmission', 'fuelType']

Xlabel = X

for i in columns:
    Xlabel[i] = encode.fit_transform(Xlabel[i])

Xlabel.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

numeric_columns = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
X_one_hot_encoded[numeric_columns] = scaler.fit_transform(X_one_hot_encoded[numeric_columns])
```

```python
X_one_hot_encoded
```

```python
Xlabel.columns
```

```python
Xlabel_columns = ['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']

Xlabel[Xlabel_columns] = scaler.fit_transform(Xlabel[Xlabel_columns])
```

```python
Xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
X_one_hot_train, X_one_hot_test, y_one_hot_train, y_one_hot_test = train_test_split(X_one_hot_encoded, y, test_size=0.2, random_state=42)
model = LinearRegression()
model.fit(X_one_hot_train, y_one_hot_train)
```

```python
y_one_hot_pred = model.predict(X_one_hot_test)
```

```python
r2 = r2_score(y_one_hot_test, y_one_hot_pred)
r2
```

```python
n = X_one_hot_test.shape[0]  # number of observations
p = X_one_hot_test.shape[1]  # number of predictors
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adjusted_r2
```

```python
# for the label encoded data
Xlabel_train, Xlabel_test, ylabel_train, ylabel_test = train_test_split(Xlabel, y, test_size=0.2, random_state=42)
model = LinearRegression()
model.fit(Xlabel_train, ylabel_train)
```

```python
ylabel_pred = model.predict(Xlabel_test)
```

```python
r2 = r2_score(ylabel_test, ylabel_pred)
r2
```
